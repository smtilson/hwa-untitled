# Tournament Presentation Demo

This notebook demonstrates the display functions in `tournament.presentation` for visualizing tournament results.

```python
from random import Random
from tournament import (
    make_players,
    run_tournament,
    get_pairing,
    display_round,
    display_tournament_summary,
    display_stacked_rounds,
    display_match_details,
    display_player_performance,
    display_skill_statistics,
    display_metrics,
    display_metrics_comparison,
)
```

## Generate a 6-round tournament with 8 players

We'll use the "adjacent" pairing strategy, which pairs players within the same record group in order.

```python
# Generate 8 players with varying skill levels
rng = Random(42)
players = make_players(8, rng)

# Run a 6-round tournament with adjacent pairing
tournament = run_tournament(
    players,
    n_rounds=6,
    pairing=get_pairing("adjacent"),
    rng=rng
)

print(f"Tournament completed: {len(tournament.players)} players, {len(tournament.rounds)} rounds")
```

## Display Round-by-Round Results

The `display_round` function shows each match in a round with:
- Player pairing (names and IDs)
- Skill levels
- Previous record (before this round)
- Agent totals for and against

```python
# Display Round 1
print(display_round(tournament, round_number=1))
```

```python
# Display Round 3
print(display_round(tournament, round_number=3))
```

```python
# Display Round 6 (final round)
print(display_round(tournament, round_number=6))
```

## Custom Fields with extra_fields

You can add custom fields to the round display using the `extra_fields` callback.

```python
def custom_fields(pid, player, record):
    """Add custom fields to display."""
    return {
        "agent_total_ratio": f"{record.agent_total_ratio:.3f}",
        "matches_played": record.matches_played
    }

print(display_round(tournament, round_number=2, extra_fields=custom_fields))
```

## Tournament Summary

The `display_tournament_summary` function shows final standings with records and agent differentials.

```python
print(display_tournament_summary(tournament))
```

## Stacked Rounds View

The `display_stacked_rounds` function shows rounds stacked side-by-side:
- Each row is a player
- Each column is a round
- Shows which group the player is in at each stage

This view makes it easy to track player progression through the tournament.

```python
print(display_stacked_rounds(tournament, show_groups=True))
```

You can also hide group information to see just the records:

```python
print(display_stacked_rounds(tournament, show_groups=False))
```

## Match Details

The `display_match_details` function shows detailed information about a specific match, including game-by-game results.

```python
# Show details of the first match in Round 1
print(display_match_details(tournament, round_number=1, match_index=0))
```

## Player Performance

The `display_player_performance` function shows a player's performance over the course of the tournament, including:
- Opponent PID for each round
- Opponent's record before the match
- Opponent's total agents scored up to that point
- Opponent's agents taken (agents against)
- Opponent's agent differential
- Opponent's agent ratio
- The group the player is in that round

```python
# Show performance for player 0 (the first player)
print(display_player_performance(tournament, player_pid=0))
```

## Skill Statistics

The `display_skill_statistics` function shows basic statistics about the skill levels of players in the tournament: mean, median, standard deviation, min, max, and range.

```python
print(display_skill_statistics(tournament))
```

## Evaluation Metrics

The `display_metrics` function shows quality metrics for evaluating the pairing algorithm:
- **Mean skill gap** — average absolute skill difference between paired players (lower = better matched)
- **Standings-skill correlation** — correlation between final standings and true skill (+1 = perfect, 0 = none)
- **Rematch count** — how many times any pair was matched more than once (lower = better)

```python
print(display_metrics(tournament))
```

## Comparison: Different Pairing Strategies

Let's compare how different pairing strategies affect tournament outcomes.

```python
# Generate a new tournament with "fold" pairing
rng2 = Random(123)
players2 = make_players(8, rng2)

tournament_fold = run_tournament(
    players2,
    n_rounds=6,
    pairing=get_pairing("fold"),
    rng=rng2
)

print("=== Fold Pairing Tournament Summary ===")
print(display_tournament_summary(tournament_fold))
```

```python
# Generate a tournament with "strong_weak" pairing
rng3 = Random(456)
players3 = make_players(8, rng3)

tournament_sw = run_tournament(
    players3,
    n_rounds=6,
    pairing=get_pairing("strong_weak"),
    rng=rng3
)

print("=== Strong-Weak Pairing Tournament Summary ===")
print(display_tournament_summary(tournament_sw))
```

```python
## Summary

The presentation module provides five main display functions:

1. **`display_round`** - Shows detailed information for a specific round, including pairings, skills, previous records, and agent totals. Supports custom fields via the `extra_fields` callback.

2. **`display_tournament_summary`** - Shows final standings with rankings, records, agent differentials, and ratios.

3. **`display_stacked_rounds`** - Shows a matrix view with players as rows and rounds as columns, making it easy to track player progression through the tournament. Can show group information or just records.

4. **`display_match_details`** - Shows detailed game-by-game results for a specific match.

5. **`display_player_performance`** - Shows a player's performance over the course of the tournament, including opponent information, opponent records, agent totals, and group information for each round.

These functions are useful for:
- Analyzing tournament progression
- Comparing different pairing strategies
- Debugging tournament simulations
- Presenting results to stakeholders
```
